```python
import numpy as np
import pandas as pd
import nltk
import string
nltk.download('names')
from nltk.corpus import names
```

```python
names.words('female.txt')
```

```python
# i for i in names.words('gender.txt')
male_names = [(name.lower(), 'male')for name in names.words('male.txt')]
female_names = [(name.lower(), 'female')for name in names.words('female.txt')]

data = male_names + female_names
data
```

```python
np.random.shuffle(data)
data
```

```python
X = [name for name, gender in data]
y = [gender for name, gender in data]
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_encoded = le.fit_transform(y) # male=1, female=0
```

```python
y_encoded
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer(char_level=True)
tokenizer.fit_on_texts(X)
X_seq = tokenizer.texts_to_sequences(X)
X_padded = pad_sequences(X_seq, maxlen=15) #pad/truncate to 15 characters
```

```python
X_padded
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_padded, y_encoded, test_size=0.3, random_state=42)
```

```python
# Step 6: Build RNN Model
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

vocab_size = len(tokenizer.word_index) + 1  # char-level vocab
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=8, input_length=15),
    SimpleRNN(32),
    Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=20, validation_data=(X_test, y_test))
```

```python
from sklearn.metrics import classification_report

y_pred_prob = model.predict(X_test)
y_pred = (y_pred_prob > 0.5).astype(int)
print("RNN Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression
X_count = tokenizer.texts_to_matrix(X, mode='binary')
X_train_lr, X_test_lr, y_train_lr, y_test_lr = train_test_split(X_count, y_encoded, test_size=0.2, random_state=42)

logreg = LogisticRegression(max_iter=1000)
logreg.fit(X_train_lr, y_train_lr)
y_pred_lr = logreg.predict(X_test_lr)

print("\nLogistic Regression Classification Report:")
print(classification_report(y_test_lr, y_pred_lr))
```

```python
def predict_gender(name):
  seq = tokenizer.texts_to_sequences([name.lower()])
  padded_seq = pad_sequences(seq, maxlen=15)
  prediction = model.predict(padded_seq)[0][0]
  gender = 'Male' if prediction > 0.5 else 'Female'
  print(f"{name} is likely {gender} ({prediction:.2f})")
```

```python
print(predict_gender('Anukriti'),
predict_gender('Nikhil'))
```
